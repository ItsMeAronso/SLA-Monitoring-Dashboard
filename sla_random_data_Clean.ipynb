```python
import pandas as pd
import random
from datetime import datetime, timedelta

customers_data = pd.DataFrame({
    'Customer Name': [f"Customer_{i}" for i in range(1, 51)]
})

countries = ['USA', 'UK', 'Germany', 'France', 'Italy', 'Spain']
addresses_data = pd.DataFrame({
    'Id': range(100, 150),
    'Country': [random.choice(countries) for _ in range(50)]
})

orders_data = []
base_date = datetime.now() - timedelta(days=90)
couriers = ['FedEx, Express', 'DHL, Standard', 'UPS, Freight', 'Israel Post, Local']

for i in range(637):
    creation_date = base_date + timedelta(days=random.randint(0, 90))
    deliver_on = creation_date + timedelta(days=random.randint(3, 7))
    
    rand_sla = random.random()
    if rand_sla < 0.7:
        close_date = deliver_on - timedelta(days=random.randint(0, 2))
    elif rand_sla < 0.9:
        close_date = deliver_on + timedelta(days=random.randint(1, 3))
    else:
        close_date = deliver_on + timedelta(days=random.randint(4, 12))
        
    orders_data.append({
        'Reference': f"ORD-{10000 + i}",
        'Courier': random.choices(couriers, weights=[0.45, 0.35, 0.15, 0.05])[0],
        'Creation date': creation_date.strftime("%d/%m/%Y"),
        'Deliver On': deliver_on.strftime("%d/%m/%Y"),
        'Close date': close_date.strftime("%d/%m/%Y"),
        'Created by': random.choice(['System_API', 'Manual_Entry']),
        'Customer': random.choice(customers_data['Customer Name']),
        'Address Id': random.choice(addresses_data['Id'])
    })

orders = pd.DataFrame(orders_data)
customers = customers_data
addresses = addresses_data

if 'Created by' in orders.columns:
    orders = orders[orders['Created by'] == 'System_API']

if 'Courier' in orders.columns:
    orders['Courier'] = orders['Courier'].astype(str).str.split(',').str[0]

merged_data = pd.merge(orders, customers, left_on='Customer', right_on='Customer Name', how='left')
master_data = pd.merge(merged_data, addresses, left_on='Address Id', right_on='Id', how='left')

final_columns = ['Reference', 'Courier', 'Country', 'Creation date', 'Deliver On', 'Close date']
master_data = master_data[final_columns]

output_filename = "Fake_Master_Data_Realistic.csv"
master_data.to_csv(output_filename, index=False)

print(f"File '{output_filename}' created successfully with {len(master_data)} records!")
```
